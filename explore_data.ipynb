```python
import pandas as pd

# 1. Load your catalog CSV from the data folder
catalog_path = './data/Skyserver.csv'  # update filename if needed
catalog = pd.read_csv(catalog_path, comment='#')

# 2. Define valid major spectral classes
VALID_CLASSES = set('OBAFGKM')

def major_spectral_class(subclass):
    if subclass is None:
        return None
    s = str(subclass).strip().upper()
    if not s:
        return None
    first = s[0]
    return first if first in VALID_CLASSES else None

# 3. Map subclasses to the primary target letter and drop invalid rows
catalog['target'] = catalog['subClass'].apply(major_spectral_class)
catalog = catalog.dropna(subset=['target'])

# 4. Inspect the distribution of your stellar classes
print(catalog['target'].value_counts())
```
